### Loading the Clinical Trial Data

```python
import pandas as pd

url = "https://vincentarelbundock.github.io/Rdatasets/csv/survival/pbc.csv"
pbc = pd.read_csv(url)
```

### Non-trivial transformation/Analysis: finding the mean age for each sex in the trial.

```python
pbc1 = pbc.groupby('sex').mean()['age']
pbc1
```

### Visualization

```python
pbcfemale = pbc[pbc['sex'] == 'f']
pbcmale = pbc[pbc['sex'] == 'm']
```

```python
import matplotlib.pyplot as plt
from statsmodels.nonparametric.smoothers_lowess import lowess

plt.scatter('age', 'platelet', data=pbcmale)
plt.scatter('age', 'platelet', data=pbcfemale)

plt.xlabel('age')
plt.ylabel('platelet')



```
